# pdukit quickstart

Four objects: `Ensemble` (your trajectories), `Config` (settings), `Cog` (the run) and `Result` (what it found). In Colab, run `!pip install "pdukit[xtc,plot,notebook]"` first.

```python
import pdukit as pdu
print(pdu.__version__)
```

## 1. The data
Replace `pdu.demo()` with `pdu.Ensemble(["rep1.xtc", "rep2.xtc"], "top.pdb")` for your own system. The demo is a 40-residue toy protein with four planted units.

```python
ens = pdu.demo()
ens
```

## 2. The settings
Every parameter is an attribute; assignment is checked.

```python
cfg = pdu.Config.from_ppf()          # defaults; or Config.from_ppf('my.ppf')
cfg.lag = 20
cfg2 = cfg.copy()                    # independent copy
cfg2.lag = 30
(cfg.lag, cfg2.lag, cfg.source('lag'))
```

## 3. Run COG
The tutorial parameter file is tuned for the demo.

```python
from pathlib import Path
import pdukit
ppf = Path(pdukit.__file__).parent / 'params' / 'tutorial.ppf'
cfg = pdu.Config.from_ppf(ppf)
result = pdu.Cog(cfg).run(ens, out='runs/demo')
result
```
